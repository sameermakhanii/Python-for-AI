```python
import pandas as pd
import torch
import torch.nn as nn

students = pd.read_csv("students_100.csv")

x = torch.tensor(
    students[["study_hours", "phone_hours"]].values,
    dtype=torch.float32,
)

y = torch.tensor(
    students[["maths"]].values,
    dtype=torch.float32,
)

print("X Shape:", x.shape, "Y Shape:", y.shape)

torch.manual_seed(42)

model = nn.Sequential(
    nn.Linear(2, 3),
    nn.ReLU(),
    nn.Linear(3, 1),
)

loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.1)

for epoch in range(300):
    guess = model(x)
    loss = loss_fn(guess, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(f"Epoch {epoch + 1}, Loss: {loss.item():.4f}")
```

```python
model = nn.Sequential(
    nn.Linear(2, 3),
    nn.ReLU(),
    nn.Linear(3, 1),
)

loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.1)

losses = []

for epoch in range(300):
    guess = model(x)
    loss = loss_fn(guess, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    losses.append(loss.item())

    if epoch % 50 == 0:
        print(f"epoch {epoch:3}: loss = {loss.item():.4f}")

plt.plot(losses)
plt.ylim(0, 1000)
plt.title("Loss during training")
plt.xlabel("Epochs")
plt.ylabel("Loss (MSE)")
plt.show()

with torch.no_grad():
    new_student = torch.tensor([[3, 2]], dtype=torch.float32)
    print(
        "3 study hours, 2 phone hours >",
        round(model(new_student).item(), 2),
        "maths score"
    )
```
